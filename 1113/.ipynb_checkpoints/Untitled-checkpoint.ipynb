```python
import numpy as np
```

```python
A = np.array([[1,2,3,-1],[-1,-1,-2,-3]])
B = np.array([[1,2,-1],[1,3,-1],[1,-1,0],[0,-1,-1]])
A @ B
```

```python
A @ B == np.dot(A,B)
```

```python
A = np.array([[1,2,3,-1],[-1,-1,-2,-3]])
B = np.array([[1,2,-1],[1,3,-1],[1,-1,0],[0,-1,-1]])
C = A@B
# np.linalg.inv(C) 
```

<h1>정사각형이 아니기 때문에, 역행렬이 불가능 하다.</h1>

```python
A = np.array([[1,2],[-1,3]])
Result = np.linalg.inv(A) @ A
```

```python
Result < 0.001
```

```python
for i in range(2):
    for j in range(2):
        if i != j :
            print(Result[i][j] < 0.001)
        else:
            print(Result[i][j]-1 < 0.001)
```
