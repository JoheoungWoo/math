```python
# 벡터2d
a2_vec = np.array([50,2])
b2_vec = np.array([-1,3])

a3_vec = np.array([2,3,1])
b3_vec = np.array([1,-1,2])
```

```python
# magnitude 구함
a2_vec_mag = np.linalg.norm(a2_vec)
b2_vec_mag = np.linalg.norm(b2_vec)

a3_vec_mag = np.linalg.norm(a3_vec)
b3_vec_mag = np.linalg.norm(a3_vec)


a2_vec_mag, b2_vec_mag, a3_vec_mag, b3_vec_mag
```

```python
def rad_to_deg(rad):
    return rad *180.0/np.pi

def deg_to_rad(deg):
    return deg *np.pi/180.0
```

```python
# cos theta구함
cos2_theta = np.dot(a2_vec, b2_vec) / (a2_vec_mag * b2_vec_mag)
cos3_theta = np.dot(a3_vec, b3_vec) / (a3_vec_mag * b3_vec_mag)
cos2_theta, cos3_theta
```

```python
# deg 구함
rad2 = np.arccos(cos2_theta)
rad3 = np.arccos(cos3_theta)

#radina -> degree
# deg2 = np.rad2deg(rad2)
deg3 = np.rad2deg(rad3)
deg2 = rad_to_deg(rad2)
# deg3 = rad_to_deg(rad3)

deg2, deg3
```

```python
class A:
    def __init__(self):
        print("사람")
        
    def add(self,a,b):
        return a+b
    def minus(self, a,b):
        return a-b
    def mul(self, a,b):
        return a*b
    def div(self, a,b):
        return a/b
```

```python
a = A()
print(a.add(3,4))
print(a.minus(10,7))
print(a.mul(3,10))
print(a.div(10,2))

[a.add(3,4), a.minus(10,7), a.mul(3,10), a.div(10,2)]
```
